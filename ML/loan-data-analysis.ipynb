```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
loan_data = pd.read_csv('/Users/manojkumar/Documents/repository/python-programming-workbench/loan-data-set.csv')

loan_data.info()

rows, cols = loan_data.shape

print(f'rows : {rows}, col : {cols}')
```

```python
for col in loan_data.columns:
    print(f'Column : {col}, Unique Values : {loan_data[col].nunique()}')


print(loan_data['loan_status'].value_counts())
#print(loan_data['int_rate'].value_counts(normalize=True))
```

```python
fig = plt.figure(figsize=(18, 6))
spec = fig.add_gridspec(1, 2)
sec_1 = fig.add_subplot(spec[0, 0])
sec_2 = fig.add_subplot(spec[0, 1])

# adding colour prefrecences
bg_color = '#f5f5f5'
for area in [sec_1, sec_2]:
    area.set_facecolor(bg_color)

#plotting the graphs
ax = (round(loan_data['loan_status'].value_counts(normalize=True) * 100, 2)).plot(kind='bar', color=['green', 'red','blue'])
ax.bar_label(ax.containers[0])

m_values = loan_data['loan_status'].isnull().sum()

print(f'null values in loan_status : {m_values}')
```
