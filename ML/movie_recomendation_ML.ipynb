```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.model_selection as model_selection
import sklearn.linear_model as linear_model
# Cosine similarity helps find similarity between movies
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df =pd.read_csv('/Users/manojkumar/Documents/repository/python-programming-workbench/movie_recommendation_dataset.csv')
```

```python
df.describe()
```

```python
# Rows    -> Users
# Columns -> Movies
# Values  -> Ratings


movie_matrix = df.pivot_table(
    index='User',
    columns='Movie',
    values='Rating'
)


print("\nUser-Movie Matrix:\n")
print(movie_matrix)
```

```python
# Replace NaN values with 0


movie_matrix = movie_matrix.fillna(0)


print("\nMatrix After Filling Missing Values:\n")
print(movie_matrix)
```

```python
# Transpose matrix to compare movies


movie_similarity = cosine_similarity(movie_matrix.T)


# Convert similarity array into DataFrame


similarity_df = pd.DataFrame(
    movie_similarity,
    index=movie_matrix.columns,
    columns=movie_matrix.columns
)


print("\nMovie Similarity Matrix:\n")
print(similarity_df)
```
