# PROMPTS AND PROMPT TEMPLATES 

```python
from langchain_core.prompts import PromptTemplate

# creating a prompt template with input variables
prompt_template = PromptTemplate.from_template(
    "Tell me a {adjective} joke about {content}."
)

# PromptTemplate this is a langchain class for creating resuable prompts with placeholders 
# here placeholders are adjective and content basically the input variables 
# PromptTemplate.from_template - this is basically creating the template from the given text and storing in the variable  prompt_template
```

```python
prompt_template.input_variables 
```

```python
prompt  = prompt_template.format(adjective="funny",content="data science")
print(type(prompt))
print(prompt)

#  .format() replaces the placeholders  in the template with what ever user inputs 
 # prompt_template IS CALLING THE .format() method to fill the placeholders and create the full prompt 
# result is just a plain string 
```

```python
prompt  = prompt_template.format(adjective="bad",content="ML science")
print(type(prompt))
print(prompt)
```

```python
prompt  = prompt_template.format(adjective="nice",content="cooking")
print(type(prompt))
print(prompt)
```

```python
prompt  = prompt_template.format_prompt(adjective="crazy",content="space")
print(type(prompt))
print(prompt)
# format_prompt is also used to replace the placeholders with user values  . 
# but here instead of a python str it returns a promptvalue object - this can be later on passsed to the model 
```

```python
prompt  = prompt_template.format_prompt(adjective="crazy",content="space").to_messages()
print(type(prompt))
print(prompt)
# to_messages() helps create  a list of langchain message objects 
```

# chat prompt template  -  from messages()

```python
from langchain_core.prompts import ChatPromptTemplate
chat_template = ChatPromptTemplate.from_messages(
    [
        ("system","you are a helpful AI Bot. Your name is {name}."),
        ("human","Hello, how are you doing today?"),
        ("ai","I'm doing well, thanks!"),
        ("human","{user_input}"),        
    ]
    
)

# ChatPromptTemplate  this is used to create a multiple messages with specific roles - user, system , ai etc
# the .from_messages is helping to create this list of messages 

# basically both are working together to creae that list of messages for conversations by defining roles  
# creaing a structured conversation so that when it goes to LLM , llm knows who is talking what 
```

# DESIGN A PROMPT TEMPLATE BY USING HUMANMESSAGEPROMPT TEMPLATE, SYSTEMMESSAGEPROMPT TEMPLATE, AIMESSAGEPROMPT TEMPLATE

```python
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate,AIMessagePromptTemplate 
```

```python
#system template
system_prompt_template = SystemMessagePromptTemplate.from_template("You are a helpful AI BOT. YOur name is  {name}")
#human template
human_prompt_template = HumanMessagePromptTemplate.from_template("Hello How are you doing ")
#ai template
ai_prompt_template = AIMessagePromptTemplate.from_template("Hey thanks, am doing great")
#human template
human_prompt_template_input = HumanMessagePromptTemplate.from_template("{user_input} ")

# THIS CODE IS BASICALLY CREATING INDIVIDUAL TEMPLATES FOR EACH ROLE INSTEAD OF PUTTING TOGEHTER LIKE WE DID IN EARLIER CELL  
#ChatPromptTemplate.from_messages
# .from_template IS CREATING A TEMPLATE FOR EACH SPECIFIC ROLE 
```

```python
# compile the whole chat prompt
chat_template = ChatPromptTemplate.from_messages([system_prompt_template,human_prompt_template,ai_prompt_template,human_prompt_template_input])
chat_template.format_prompt(name="EVA", user_input="What is your name").to_messages()
```
